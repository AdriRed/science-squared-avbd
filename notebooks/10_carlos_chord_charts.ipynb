```python
import numpy as np
import pandas as pd
from pathlib import Path
import holoviews as hv
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
from bokeh.models import ColumnDataSource, HoverTool, TapTool, CustomJS
from bokeh.palettes import Category20

hv.extension('bokeh')
output_notebook()

data_dir = Path('../data')
cache_dir = data_dir / 'cache'
external_dir = data_dir / 'external'
processed_dir = data_dir / 'processed'
plots_dir = Path('../plots/carlos')
```

```python
df = pd.read_csv(processed_dir / 'links_all.csv')
links = df.rename(columns={'institution_a': 'source',
                           'institution_b': 'target',
                           'weight': 'value'})
links
```

```python
all_nodes = sorted(pd.unique(links[['source', 'target']].values.ravel()))
nodes = hv.Dataset(pd.DataFrame({'name': all_nodes}), 'name')

# one color per university
palette = Category20[20]
colors = {name: palette[i % 20] for i, name in enumerate(all_nodes)}

# HoloViews computes the geometry only
chord = hv.Chord((links, nodes))
edge_paths = chord.edgepaths.split()      # one path per row of `links`
node_df = chord.nodes.dframe()
node_df['hex'] = node_df['name'].map(colors)
```

```python
def make_edges(n_segments=25):
    """Cut every chord into n_segments pieces and color each one."""
    segments = []
    for row, path in zip(links.itertuples(index=False), edge_paths):
        xy = path.array()
        if len(xy) == 0:
            continue
        # an edge is made of several curves separated by NaN rows
        gaps = np.where(np.isnan(xy[:, 0]))[0]
        curves = [c[~np.isnan(c[:, 0])] for c in np.split(xy, gaps)]
        curves = [c for c in curves if len(c) > 1]

        for k in range(n_segments):
            pieces = []
            for c in curves:
                cuts = np.linspace(0, len(c) - 1, n_segments + 1).round().astype(int)
                pieces += [c[cuts[k]:cuts[k+1] + 1], [[np.nan, np.nan]]]
            seg = np.concatenate(pieces[:-1])
            segments.append({
                'x': seg[:, 0], 'y': seg[:, 1],
                'source': row.source, 'target': row.target, 'value': row.value,
                'hex': blend(colors[row.source], colors[row.target],
                             k / max(n_segments - 1, 1)),
            })

    hover = HoverTool(tooltips=[('Source', '@source'),
                                ('Target', '@target'),
                                ('Weight', '@value')])
    return hv.Path(segments, vdims=['source', 'target', 'value', 'hex']).opts(
        color=hv.dim('hex'), line_width=1, line_alpha=0.7, tools=[hover])
```

```python
def make_chord(csv_path, size=800):
    """Two-color chord diagram from a CSV with institution_a, institution_b, weight."""
    links = pd.read_csv(csv_path).rename(columns={'institution_a': 'source',
                                                  'institution_b': 'target',
                                                  'weight': 'value'})
    links = links[['source', 'target', 'value']]

    names = sorted(pd.unique(links[['source', 'target']].values.ravel()))
    colors = {n: Category20[20][i % 20] for i, n in enumerate(names)}

    # HoloViews computes the geometry only
    chord = hv.Chord((links, hv.Dataset(pd.DataFrame({'name': names}), 'name')))

    # split every curve at its midpoint: half "a" (source) and half "b" (target)
    data = {k: [] for k in ['xa', 'ya', 'xb', 'yb']}
    for path in chord.edgepaths.split():
        xy = path.array()
        gaps = np.where(np.isnan(xy[:, 0]))[0]
        curves = [c[~np.isnan(c[:, 0])] for c in np.split(xy, gaps)]
        curves = [c for c in curves if len(c) > 1]
        for half, part in (('a', 0), ('b', 1)):
            pieces = []
            for c in curves:
                mid = len(c) // 2
                piece = c[:mid + 1] if part == 0 else c[mid:]
                pieces += [piece, [[np.nan, np.nan]]]
            seg = np.concatenate(pieces[:-1]) if pieces else np.empty((0, 2))
            data['x' + half].append(seg[:, 0].tolist())
            data['y' + half].append(seg[:, 1].tolist())
    data.update(source=links['source'].tolist(), target=links['target'].tolist(),
                value=links['value'].tolist(),
                ca=links['source'].map(colors).tolist(),
                cb=links['target'].map(colors).tolist())
    edges = ColumnDataSource(data)

    p = figure(width=size, height=size, x_range=(-1.5, 1.5), y_range=(-1.5, 1.5),
               tools='reset,save')
    p.axis.visible = False
    p.grid.visible = False
    p.outline_line_color = None

    # both halves share one data source, so selecting a chord selects both halves
    halves = []
    for h in 'ab':
        r = p.multi_line('x' + h, 'y' + h, source=edges, line_color='c' + h,
                         line_width=1, line_alpha=0.6)
        r.selection_glyph = r.glyph.clone(line_alpha=1, line_width=1.5)
        r.nonselection_glyph = r.glyph.clone(line_alpha=0.04)
        halves.append(r)

    # hover: tooltip + highlight of the whole chord (both halves)
    for r in halves:
        marked = ColumnDataSource({k: [] for k in data})
        for h in 'ab':
            p.multi_line('x' + h, 'y' + h, source=marked, line_color='c' + h,
                         line_width=2, line_alpha=1)
        callback = CustomJS(args=dict(edges=edges, marked=marked), code="""
            const rows = cb_data.index.indices
            const out = {}
            for (const key in edges.data)
                out[key] = rows.map(i => edges.data[key][i])
            marked.data = out
        """)
        p.add_tools(HoverTool(renderers=[r], callback=callback, line_policy='nearest',
                              tooltips=[('Source', '@source'),
                                        ('Target', '@target'),
                                        ('Weight', '@value')]))

    # university dots and labels
    nodes = chord.nodes.dframe()
    angle = np.arctan2(nodes['y'], nodes['x'])
    left = nodes['x'] < 0
    nodes['color'] = nodes['name'].map(colors)
    nodes['lx'] = 1.08 * np.cos(angle)
    nodes['ly'] = 1.08 * np.sin(angle)
    nodes['angle'] = np.where(left, angle + np.pi, angle)
    nodes['align'] = np.where(left, 'right', 'left')
    node_source = ColumnDataSource(nodes)

    dots = p.scatter('x', 'y', source=node_source, size=15,
                     fill_color='color', line_color='black')
    labels = p.text('lx', 'ly', text='name', source=node_source, angle='angle',
                    text_align='align', text_baseline='middle', text_font_size='8pt')
    # keep dots and labels readable while something is selected
    for r in (dots, labels):
        r.nonselection_glyph = r.glyph.clone()
    p.add_tools(HoverTool(renderers=[dots], tooltips=[('University', '@name')]))

    # click a chord to isolate it; click a university to show all its chords
    tap = TapTool(renderers=halves + [dots])
    p.add_tools(tap)
    p.toolbar.active_tap = tap
    select_node = CustomJS(args=dict(edges=edges, nodes=node_source), code="""
        // wait until the click has updated both chords and dots
        setTimeout(() => {
            const picked = nodes.selected.indices.map(i => nodes.data.name[i])
            if (picked.length == 0) return
            const rows = []
            edges.data.source.forEach((s, i) => {
                if (picked.includes(s) || picked.includes(edges.data.target[i])) rows.push(i)
            })
            if (rows.join() != edges.selected.indices.join())
                edges.selected.indices = rows
        }, 0)
    """)
    node_source.selected.js_on_change('indices', select_node)
    # a click on a dot also touches the chords under it, so the dot takes priority
    edges.selected.js_on_change('indices', select_node)

    return p
```
